## 1.1 Storing answers — variables

A **variable** is a name that holds a value. The `=` does not mean "equals" — it means
**put this into**.

````python
year_birth = 1957
education  = "Bachelor"
income     = 58138
kids       = 0
responded  = True          # our records show they accepted the last offer

print(education)
````

Two rules for naming: letters, digits and underscores only — `year_birth` works,
`year-birth` breaks — and it cannot start with a digit. Name things for what they *are*.
`x` tells nobody anything.

## 1.2 Four kinds of value

Every value in Python has a **type**. Four of them cover everything in this course.

| Type | What it is | Here |
|---|---|---|
| `int` | a whole number | `year_birth = 1957` |
| `float` | a number with a decimal point | `2.91` |
| `str` | text, always in quotes | `education = "Bachelor"` |
| `bool` | `True` or `False`, nothing else | `responded = True` |

`type()` asks Python what something is.

````python
print(type(year_birth))
print(type(education))
print(type(responded))
````

⚠️ **Quotes decide everything.** `58138` is a number you can do arithmetic with.
`"58138"` is **text that looks like a number** — and survey exports are full of them.

## 1.3 Arithmetic — how old is this person?

They were born in 1957. It is 2026.

````python
age = 2026 - year_birth     # the simple way: year only, ignore the month

print(age)
````

⚠️ **That is the simple way, and it is slightly wrong on purpose.**

If their birthday has not happened yet this year, they are still 68 — not 69. To do it properly
you would need their birth *month*, and this file only records the **year**.

So we take the simple version and say so. That is a normal, honest choice: **the data decides what
you are allowed to calculate.** Whenever you take a shortcut like this, write it down — in your
project report it belongs in the limitations section.

This happens constantly. Your survey tool exports everything as text, and the number you need to

## 1.5 How to read the red text

Three rules, and they will unblock you all semester:

1. **Read the last line first.** That is what actually went wrong.
2. The **line number** tells you where to look.
3. Everything in between is the path Python took to get there. Ignore it for now.

The last line says `TypeError: can only concatenate str (not "int") to str`. In plain English:
*you asked me to join text to a number, and I will not guess which you meant.*

## 1.6 Converting — and **you** decide what you meant

There are two ways to "fix" that line, and they give different answers because they mean
different things.

````python
print(int(income_text) + 1000)     # 59138  -> you meant arithmetic
print(income_text + "1000")        # 581381000 -> you meant joining text
````

The conversion functions you will actually use:

| Function | Turns something into | Example |
|---|---|---|
| `int()` | a whole number | `int("58138")` → `58138` |
| `float()` | a decimal number | `float("2.5")` → `2.5` |
| `str()` | text | `str(1957)` → `"1957"` |
| `round()` | fewer decimal places | `round(2.9117, 2)` → `2.91` |

⚠️ Division **always** gives a float: `10 / 2` is `5.0`, not `5`. So your percentages will have
decimals whether you asked for them or not.

````python
income = int(income_text)
per_person = income / (kids + 1)

print(per_person)
print(round(per_person, 2))
````

## 1.7 Seeing more than one thing

A cell shows only its **last** line. That is why `print()` matters in a notebook — and `round()`,
so the decimals do not run off the page.

For money, one extra trick: an **f-string**. The `f` before the quote lets you drop a value into
text, and `:,` adds the thousands separators.

````python
print(age)
print(round(per_person, 2))
print(f"Income: {income:,} HKD")
````

# Part 3 · The dot

Look at the two lines you just ran:

```python
pd.read_csv('marketing_campaign.csv')
df.head()
```

They look the same. They are **different kinds of thing**, and so is `df.shape`, which you are
about to meet. Four kinds in total:

| Kind | Looks like | What the dot is doing |
|---|---|---|
| **Built-in function** | `print(x)` · `len(x)` · `int(x)` | No dot at all. It comes with Python, and the thing goes *inside* the brackets |
| **Module function** | `pd.read_csv(...)` | The dot says **where to find it** — inside the package called `pd` |
| **Method** | `df.head()` · `df.tail()` | The dot says **this object — do this to yourself** |
| **Attribute** | `df.shape` · `df.columns` | The dot asks for something the object **already is**. No brackets |

## 3.1 Brackets or no brackets?

This is the part that trips everybody.

- **Brackets** → something the object **does**. An action. It has to run.
- **No brackets** → something the object **is**. A fact to read off.

Run this and compare the two.

````python
print(df.shape)     # ATTRIBUTE - no brackets. (rows, columns)
print(df.head())    # METHOD    - brackets. it goes and fetches the rows
````

Now get it wrong on purpose, so you recognise both mistakes when they happen to you later.

**Mistake 1 — brackets on an attribute.** Run it and read the last line.

````python
df.shape()
````

`TypeError: 'tuple' object is not callable`. In plain English: **`df.shape` is already an
answer, and you cannot run an answer.**

**Mistake 2 — no brackets on a method.** This one does not even error, which makes it worse.

````python
df.head
````

`<bound method NDFrame.head of ...>` — Python is showing you **the method itself**, not its
result. You handed over the recipe instead of the meal. Add the brackets and it cooks.

⚠️ **Whenever you see `<bound method ...>` in your output, you have forgotten a pair of brackets.**
It is the most common beginner bug in pandas, and now you will recognise it on sight.

Methods are not only for tables. Every type has its own — a piece of text can do things to itself
as well.

````python
word = "Bachelor"

print(len(word))       # built-in FUNCTION - word goes inside the brackets
print(word.upper())    # METHOD - word does something to itself
print(word.lower())
````

# Part 5 · The other kind of "type"

`info()` told you `int64`, `float64` and `object` (or `str`). That is **how the computer stores
it**. Useful — but it will not tell you which analysis to run.

For that you need the **scale of measurement**, and the easiest way to hold it is: *what am I
allowed to do with this column?*

| Scale | You can… | Example here | Why not more |
|---|---|---|---|
| **Nominal** | **count** them | `Marital_Status` | "Married minus Single" is nonsense. There is no order |
| **Ordinal** | count, and **rank** them | `Education` | PhD is above Master — but the gap Bachelor→Master is not the same size as Master→PhD, so you cannot subtract |
| **Interval** | count, rank, **subtract** | `Year_Birth` | 1980−1970 is a real 10 years. But year 0 is not "no time", so "1980 is twice 990" is meaningless |
| **Ratio** | count, rank, subtract, **divide** | `Income` | Zero income really is no income — so "twice as much" finally means something |

Each row can do everything the row above can, plus one more thing. That is the whole idea.

Nominal and ordinal are **discrete**. Interval and ratio are **continuous** — and those two words
are what decides your statistical test.

*Why `Year_Birth` is interval, not ratio:* you subtracted two years to get an age in Part 1, which
is allowed. But nobody born in 1980 is "twice" anyone born in 990 — there is no true zero, so no
dividing.

## 5.1 Two columns that will catch you

**`Kidhome` — a number that is not continuous.** Python says `int64`. The scale is ratio, since
zero children is a real zero. But nobody has 1.4 children, so it is **discrete**, and in an
analysis it behaves like a category.

**`Response` — a number that is really a name.** Stored as 0 and 1. Python will happily give you
its mean. But 0 and 1 mean *no* and *yes*, so it is **nominal** — and that "average" is really a
**response rate**.

````python
print(df['Response'].mean())     # 0.1518... -> not an average. A rate: 15.2% said yes.
````

**Python genuinely cannot tell you either of these.** `dtype` says `int64` for both — pandas has
no idea whether a number is a count, a category, or a measurement.

An AI often *will* get it right, and you should expect it to: it reads the column **name**, and
`Kidhome` and `Response` are informative names. But that is a guess made from the name, and it
fails exactly where your own data will be weakest — a survey export whose columns are called
`Q7_3` and `V12` gives it nothing to guess from.

So the judgement stays yours. **You are the only one who knows what the question actually asked.**
